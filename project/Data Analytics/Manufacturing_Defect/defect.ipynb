## Import

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://raw.githubusercontent.com/deristnoor/python/refs/heads/master/project/Data%20Analytics/Manufacturing_Defect/datasets/defects_data.csv"
df = pd.read_csv(url, parse_dates=["defect_date"])
df.head()
```

```python
df[df.defect_date == '2024-01-01']
```

```python
df.describe()
```

```python
df[df.isnull() == True].count()
```

```python
df.dtypes
```

```python
df['month'] = df["defect_date"].dt.strftime("%b%Y")
```

```python
df_sorted = df.sort_values("defect_date")
```

```python
fig, axes = plt.subplots(2,3, figsize=(20,10))

sns.lineplot(data=df_sorted, x='month', y='repair_cost', sort=False, ax=axes[0,0])
axes[0,0].set_title("date vs cost")

sns.lineplot(data=df_sorted, x="month", y='repair_cost', hue="defect_location", ax=axes[0,1])
axes[0,1].set_title("date vs location")

sns.lineplot(data=df_sorted, x='month', y='repair_cost', hue="defect_type", ax=axes[0,2])
axes[0,2].set_title("date vs type")

sns.lineplot(data=df_sorted, x="month", y='repair_cost', hue="severity", ax=axes[1,0])
axes[1,0].set_title("date vs severity")

sns.lineplot(data=df_sorted, x='month', y='repair_cost', hue="inspection_method", ax=axes[1,1])
axes[1,1].set_title("date vs method")

# sns.histplot(data=df, x='inspection_method')

plt.show()
```
